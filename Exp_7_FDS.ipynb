```python
from google.colab import files

uploaded = files.upload()
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Get the filename from the uploaded dictionary keys
filename = list(uploaded.keys())[0] # Assuming only one file was uploaded

df = pd.read_csv(filename)
df.head()

print("\nDataset Shape")
print(df.shape)
```

Check Missing Values

```python
print("\nMissing Values")
print(df.isnull().sum())
```

Handle Missing Values

```python
# Fill missing Age with Mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Fill missing Embarked with Mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Removed the line below as 'Cabin' column does not exist:
# df['Cabin'].fillna("Unknown", inplace=True)

print("\nMissing Values After Cleaning")
print(df.isnull().sum())
```

Remove Duplicate Records

```python
print("\nDuplicate Records:", df.duplicated().sum())

df = df.drop_duplicates()

print("Dataset Shape After Removing Duplicates")
print(df.shape)
```

Remove Unnecessary Columns

```python
df = df.drop(['PassengerId'], axis=1)

print("\nColumns After Dropping")
print(df.columns)
```

Rename Columns

```python
df.rename(columns={
    'Pclass':'Passenger_Class',
    'Sex':'Gender'
}, inplace=True)

print("\nRenamed Columns")
print(df.columns)
```

Check Data Types

```python
print("\nData Types")
print(df.dtypes)
```

Detect Outliers (Fare)

```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)

IQR = Q3 - Q1

Lower = Q1 - 1.5 * IQR
Upper = Q3 + 1.5 * IQR

outliers = df[(df['Fare'] < Lower) | (df['Fare'] > Upper)]

print("\nNumber of Outliers:", len(outliers))
```

Remove Outliers

```python
df = df[(df['Fare'] >= Lower) &
        (df['Fare'] <= Upper)]

print("Dataset Shape After Removing Outliers")
print(df.shape)
```

Check Missing Values Again

```python
print("\nFinal Missing Values")
print(df.isnull().sum())
```

Save Cleaned Dataset

```python
df.to_csv("Titanic_Cleaned.csv", index=False)

print("\nCleaned Dataset Saved Successfully")
```
