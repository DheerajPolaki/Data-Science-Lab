```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/Automobile_data.csv")
temp = df.replace('?', np.nan)

cols = [
'wheel-base','length','width','height',
'curb-weight','engine-size',
'compression-ratio',
'horsepower',
'city-mpg',
'highway-mpg',
'price'
]

for c in cols:
    temp[c] = temp[c].astype(float)

plt.figure(figsize=(10,8))

sns.heatmap(
    temp[cols].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Matrix")
plt.show()
```

```python
temp = df.replace('?', np.nan)

temp['wheel-base'] = temp['wheel-base'].astype(float)
temp['length'] = temp['length'].astype(float)

row1 = temp.iloc[0][['wheel-base','length']]
row2 = temp.iloc[10][['wheel-base','length']]

distance = np.sqrt(
    (row1['wheel-base'] - row2['wheel-base'])**2 +
    (row1['length'] - row2['length'])**2
)

print("Euclidean Distance:", distance)
```

```python
manhattan = abs(row1['wheel-base'] - row2['wheel-base']) + \
            abs(row1['length'] - row2['length'])

print("Manhattan Distance:", manhattan)
```

```python
minkowski = (
    abs(row1['wheel-base'] - row2['wheel-base'])**3 +
    abs(row1['length'] - row2['length'])**3
) ** (1/3)

print("Minkowski Distance:", minkowski)
```

```python
A = np.array(row1)
B = np.array(row2)

cosine_similarity = np.dot(A,B) / (np.linalg.norm(A)*np.linalg.norm(B))

cosine_dissimilarity = 1 - cosine_similarity

print("Cosine Dissimilarity")
print(cosine_dissimilarity)
```
