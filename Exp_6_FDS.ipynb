Step 1: Create DataFrames

```python
import pandas as pd

student = pd.DataFrame({
    'Student_ID':[101,102,103,104],
    'Name':['Alice','Bob','Charlie','David'],
    'Branch':['CSE','ECE','EEE','ME']
})

marks = pd.DataFrame({
    'Student_ID':[101,102,103,105],
    'Marks':[88,75,92,80]
})

attendance = pd.DataFrame({
    'Student_ID':[101,102,103,104],
    'Attendance':[95,90,85,88]
})

print("Student DataFrame")
print(student)

print("\nMarks DataFrame")
print(marks)

print("\nAttendance DataFrame")
print(attendance)
```

Step 2: Merge DataFrames (Inner Join)

```python
print("\nInner Merge")

inner_merge = pd.merge(student, marks, on='Student_ID')

print(inner_merge)
```

Left Merge

```python
print("\nLeft Merge")

left_merge = pd.merge(student, marks,
                      on='Student_ID',
                      how='left')

print(left_merge)
```

Right Merge

```python
right_merge = pd.merge(student, marks,
                       on='Student_ID',
                       how='right')

print(right_merge)
```

Outer Merge

```python
print("\nOuter Merge")

outer_merge = pd.merge(student, marks,
                       on='Student_ID',
                       how='outer')

print(outer_merge)
```

Join DataFrames

```python
student_index = student.set_index("Student_ID")
attendance_index = attendance.set_index("Student_ID")

joined = student_index.join(attendance_index)

print("\nJoined DataFrame")
print(joined)
```

Concatenate DataFrames (Rows)

```python
student2 = pd.DataFrame({
    'Student_ID':[106,107],
    'Name':['Eva','Frank'],
    'Branch':['IT','CSE']
})

concat_rows = pd.concat([student, student2],
                        ignore_index=True)

print("\nConcatenate Rows")
print(concat_rows)
```

Concatenate Columns

```python
concat_columns = pd.concat([student, attendance['Attendance']],
                           axis=1)

print("\nConcatenate Columns")
print(concat_columns)
```

Remove Duplicates

```python
print("\nRemove Duplicates")
duplicate_df = pd.concat([student, student])

print("Before Removing")
print(duplicate_df)

duplicate_df = duplicate_df.drop_duplicates()

print("\nAfter Removing")
print(duplicate_df)
```

Sorting

```python
print("\nSorted by Name")

print(student.sort_values(by='Name'))
```

Filtering

```python
print("\nStudents from CSE")

print(student[student['Branch']=='CSE'])
```

Handling Missing Values

```python
print("\nMissing Values")

print(left_merge.isnull().sum())

left_merge['Marks'] = left_merge['Marks'].fillna(0)

print("\nAfter Filling Missing Values")

print(left_merge)
```

Rename Columns

```python
renamed = student.rename(columns={
    'Name':'Student_Name'
})

print("\nRenamed Columns")

print(renamed)
```
