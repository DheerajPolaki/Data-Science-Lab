```python
import pandas as pd
import numpy as np
```

Create DataFrames

```python
students = pd.DataFrame({
    'Student_ID':[101,102,103,104],
    'Name':['Alice','Bob','Charlie','David'],
    'Department':['CSE','ECE','CSE','EEE'],
    'Semester':[1,1,2,2],
    'Marks':[85,78,92,88]
})

attendance = pd.DataFrame({
    'Student_ID':[101,102,103,104],
    'Attendance':[95,90,87,92]
})

print("Students DataFrame")
print(students)

print("\nAttendance DataFrame")
print(attendance)
```

Hierarchical Indexing

```python
print("\nHierarchical Indexing")

multi_df = students.set_index(['Department','Semester'])

print(multi_df)

# Access data using hierarchical index
print("\nStudents from CSE Semester 1")
print(multi_df.loc[('CSE',1)])
```

Combining DataFrames (Concat)

```python
new_students = pd.DataFrame({
    'Student_ID':[105,106],
    'Name':['Eva','Frank'],
    'Department':['IT','CSE'],
    'Semester':[1,2],
    'Marks':[91,80]
})

combined = pd.concat([students,new_students],
                     ignore_index=True)

print("\nCombined DataFrame")
print(combined)
```

Merge DataFrames

```python
merged = pd.merge(students,
                  attendance,
                  on='Student_ID',
                  how='inner')

print("\nMerged DataFrame")
print(merged)
```

Join DataFrames

```python
student_index = students.set_index('Student_ID')
attendance_index = attendance.set_index('Student_ID')

joined = student_index.join(attendance_index)

print("\nJoined DataFrame")
print(joined)
```

Reshaping (Melt)

```python
melted = pd.melt(students,
                 id_vars=['Student_ID','Name'],
                 value_vars=['Marks','Semester'])

print("\nMelted Data")
print(melted)
```

Pivot Table

```python
pivot = students.pivot_table(values='Marks',
                             index='Department',
                             columns='Semester',
                             aggfunc='mean')

print("\nPivot Table")
print(pivot)
```

Stack

```python
print("\nStack")

stacked = pivot.stack()

print(stacked)
```

Unstack

```python
print("\nUnstack")

print(stacked.unstack())
```

Sorting

```python
print("\nSorted by Marks")

print(students.sort_values(by='Marks',
                           ascending=False))
```

Filtering

```python
print("\nStudents with Marks > 85")

print(students[students['Marks']>85])
```

GroupBy

```python
print("\nAverage Marks Department-wise")

print(students.groupby('Department')['Marks'].mean())
```

Handling Missing Values

```python
students.loc[2,'Marks']=np.nan

print("\nMissing Values")

print(students.isnull().sum())

students['Marks']=students['Marks'].fillna(
                    students['Marks'].mean())

print("\nAfter Filling Missing Values")

print(students)
```

Drop Duplicates

```python
duplicate_df=pd.concat([students,students])

print("\nDuplicates Removed")

print(duplicate_df.drop_duplicates())
```
